```python
#Exercise 2
import pyomo.environ as pyo

wage = [150, 152, 148, 146, 166, 176]  
min_hours = [8, 8, 8, 8, 7, 7]


available = [
    [6, 0, 6, 0, 6], 
    [0, 6, 0, 6, 0], 
    [4, 8, 4, 0, 4], 
    [5, 5, 5, 0, 5], 
    [3, 0, 3, 8, 0], 
    [0, 0, 0, 6, 2]  
]

m = pyo.ConcreteModel()
m.x = pyo.Var(range(6), range(5), domain=pyo.NonNegativeReals)
cost = 0
for i in range(6):
    for j in range(5):
        cost = cost + wage[i] * m.x[i, j]
m.obj = pyo.Objective(expr=cost, sense=pyo.minimize)

m.con = pyo.ConstraintList()


for i in range(6):
    for j in range(5):
        m.con.add(m.x[i, j] <= available[i][j])
        
for j in range(5):
    total = 0
    for i in range(6):
        total = total + m.x[i, j]
    m.con.add(total == 14)

for i in range(6):
    total = 0
    for j in range(5):
        total = total + m.x[i, j]
    m.con.add(total >= min_hours[i])

pyo.SolverFactory('cbc').solve(m)

print("Total cost:",m.obj())
for i in range(6):
    for j in range(5):
        print(f"Candidate {i + 1} Day{ j + 1}  Hours { m.x[i, j].value}")
```
